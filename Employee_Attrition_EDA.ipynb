## 4. Key Insights and Recommendations

### Key Insights:
1. Overall attrition rate is 16.12%.

2. Employees working overtime have a higher attrition rate (30.53%) than employees not working overtime (10.44%).

3. Sales has the highest departmental attrition rate (20.63%).

4. Single employees have a higher attrition rate (25.53%) than married and divorced employees.

5. Employees aged 18–25 have a relatively high attrition rate (34.78%).

6. Sales Representatives have the highest job-role attrition rate (39.76%).

7. Employees with 0–1 years at the company have a higher attrition rate (34.88%) than longer-tenured groups.

   
### Recommendations:
1. Review overtime distribution and workload.

2. Investigate employee concerns in the Sales department.

3. Strengthen support and engagement for newer employees.

4. Review job-role-specific challenges, especially for Sales Representatives.

5. Improve employee satisfaction and work-life balance.

## 5. Conclusion

This exploratory data analysis examined employee attrition patterns using the IBM HR Analytics dataset.

The analysis identified several factors associated with employee attrition, including overtime, job role, age, job satisfaction, work-life balance, and years at the company.

The findings suggest that organizations can benefit from monitoring employee workload, improving workplace satisfaction, supporting new employees, and addressing challenges in roles with higher attrition rates.

This analysis provides insights into employee attrition patterns and can help organizations make more informed workforce-management decisions.
